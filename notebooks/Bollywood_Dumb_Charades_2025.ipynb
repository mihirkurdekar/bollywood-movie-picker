```python
import pandas as pd
```

```python
movies_imdb = pd.read_csv("title.akas.tsv", sep="\t", low_memory=False)
```

```python
movies_imdb.columns
```

```python
# get only Hindi movies and types imdbDisplay
movies_bollywood = movies_imdb[(movies_imdb['language'] == 'hi') & (movies_imdb['types'] == 'imdbDisplay')]
```

```python
movies_bollywood.columns
```

```python

movies_bollywood.describe()
```

```python
#count number of unique movies by titleId
movies_bollywood['titleId'].nunique()
```
